# 1. Introduction<a id='intro'></a>
## 1.1 Background Information<a id='sec11'></a>

Singapore relies heavily on foreign labour, a transient workforce who migrate to Singapore for low-to-mid level roles: roughly 1.4 million Work Permit and S Pass holders, including migrant domestic workers [(Ministry of Manpower, 2026)](#ref12). Remittances are the non-commercial transfer of money or goods by a foreign worker to family members, friends, or communities in their home country. Foreign workers in Singapore rely on remittances for security, speed, and regularity in channelling money home, especially as their home communities may be remote, with limited access to online tools, and remittance services offer convenient cash pickups.

The U.N. Sustainable Development Goals (SDG) aim to reduce the transaction costs of migrant remittances to less than 3 per cent and eliminate remittance corridors with costs higher than 5 per cent by 2030 [(United Nations, 2026)](#ref15). These targets were reaffirmed in 2021 among the G20 targets for cross-border payments, as a global average cost of no more than 3% for sending USD 200, with no corridor above 5% [(Financial Stability Board, 2021)](#ref6). However, the global average cost of remittance still hovers around 6.4% of the total amount sent. [(World Bank, 2024b)](#ref26). Progress has been made in lowering remittance costs, but it is far from enough.

## 1.2 Motivation<a id='sec12'></a>

In 2018, US$7.1 billion was remitted from Singapore to the nine countries studied, 98.4% of its total outflow (Dataset 2, on the broad BPM6 measure). High remittance costs reduce the money reaching workers' families, adding to the pressure the Cost of Living Crisis places on low-wage foreign labourers.

The Cost of Living Crisis, in Singapore, disproportionately disadvantages foreign labourers due to their modest wages. Rising inflation reduces the purchasing power of money earned in Singapore and money received by their communities overseas. Remittance costs form a huge burden for foreign labourers, yet research on what drives them remains a small base [(Beck et al., 2022)](#ref1). Its major studies pool 119 to 365 corridors across many sending countries ([Beck & Martínez Pería, 2011](#ref2); [Beck et al., 2022](#ref1)), a wide scope. Few remittance studies draw conclusions for a single developed country as the sender, especially Singapore, and prior Singapore-based research on how low-wage migrant workers cope with the rising cost of living analyses volume but not cost of remittance [(Tay, 2023)](#ref14).

## 1.3 Goals<a id='sec13'></a>

This project aims to inform policy by showing where Singapore stands against the SDG targets, what its costs are comprised of, and which features of corridors and products are associated with lower costs. This project holds Singapore to close analogues of these standards in its analysis and recommendations to inform Singaporean and ASEAN-coordinated social policy efforts. 

## 1.4 Project Scope<a id='sec14'></a>

This study adopts the common (geographical) understanding of remittances as transfers by migrant workers to their home communities [(Migration Data Portal, n.d.)](#ref10). This is an intentional framing, as expatriate salaries are not representative of the burdens faced by migrant workers who rely on remittances. The overarching purpose of the project is to analyse the impacts of the cost of living crisis on migrant workers in the specific area of remittances, which the broader definition is unsuitable for. Datasets 2 and 5 follow the broader BPM6 definition (used in economics and statistics), which also includes compensation of employees and therefore captures expatriate salaries alongside worker transfers [(World Bank, n.d.-b)](#ref21). To stay close to the narrower definition, corridors are dropped where the receiving country is high income or in a higher income band than the sender (see 4.2) which reduces the expatriate-salary component in those datasets, making them concur well with the geographical definition.

The project is hence scoped primarily to nine corridors from Singapore: Bangladesh, China, India, Indonesia, Malaysia, Pakistan, the Philippines, Sri Lanka, and Thailand. These corridors were selected for their large migrant communities in Singapore's workforce (Dataset 3). They also have robust data, generally quarterly to half-yearly, in the Remittance Prices Worldwide dataset for nearly a decade or more. Other corridors did not have data available from reliable sources, and self-collecting data would count as an audit study requiring ethics (IRB) approval [(Committee on the Use of Human Subjects, n.d.)](#ref5).

Each (b) research question is answered against the worldwide scope: all in-scope corridors surveyed by Remittance Prices Worldwide. A more in-depth analysis occurs in the (a) sections for the nine corridors from Singapore. Sources consulted for the project proposal: [Cohen (1973)](#ref3), [Committee on the Use of Human Subjects (n.d.)](#ref5), [G20 (2011)](#ref7), [Koenker and Machado (1999)](#ref9), [Ministry of Manpower (2020)](#ref11), [World Bank (n.d.-c)](#ref22) and [World Bank (2024a)](#ref25).


# 2. Summary of Research Questions & Results<a id='summary'></a>

1. **1.a** How has the average cost of sending USD 200 from Singapore changed over 2011–2025, and do the costs today meet the SDG 10.c targets?

   **Answer:** The cost of sending USD 200 from Singapore fell by about half and has plateaued since 2022 above the 3% target; the 5% ceiling is met in all nine corridors in the UN metric.

2. **1.b** How do the level and timing of Singapore's trend compare with worldwide corridors, also in relation to the SDG 10.c targets?

   **Answer:** Singapore follows the worldwide pattern at a lower level and has made better progress towards both targets than the world.

3. **2.a** How much of the change in Singapore's remittance cost came from explicit fees versus implicit foreign exchange margins?

   **Answer:** Singapore's price of remittance fell because the direct fee of quotes fell while the FX margin rose only slightly, hence the share of the price held by FX margin doubled.

4. **2.b** Is the shift from fees to foreign exchange margins also seen worldwide, and does it survive excluding non-transparent quotes?

   **Answer:** The same pattern is evident worldwide, with a smaller rise in the margin's share; excluding non-transparent quotes leaves the result unchanged.

5. **3.a** Within Singapore's remittance market, which features of a transfer are associated with lower costs?

   **Answer:** Bank quotes cost more than non-bank quotes over the latest four quarters; worldwide, slower corridors are more expensive only because speed is acting as a proxy for bank share.

6. **3.b** Across worldwide corridors, which corridor characteristics are associated with lower costs?

   **Answer:** Bank share of quotes is the strongest factor, corridors with larger outflow and diaspora are cheaper, and digital connectivity, human development and inflation show no association with cost.

7. [UNDP Human Development Index, full time series 1990–2023](https://hdr.undp.org/data-center/documentation-and-downloads) the UNDP composite of life expectancy, schooling and income per head, one score per country per year, used as a reliably computed metric of development. [(UNDP, 2025)](#ref17)
8. [Inflation, consumer prices (annual %)](https://data.worldbank.org/indicator/FP.CPI.TOTL.ZG) — World Bank WDI annual CPI inflation for every country. [(World Bank, n.d.-a)](#ref20).

## 4.2 Data Cleaning<a id='cleaning'></a>

### 4.2.1 Correction diagnostics and case studies<a id='sec421'></a>

In 4.2.2, some corrections are made to the dataset. This is influenced by the following diagnostics ran on anomalous patterns in the data for the Singaporean datasets. 

```python
RPW_UNIFY = {"Turkey": "Türkiye", "Swaziland": "Eswatini", "Macedonia, FYR": "North Macedonia"}
for sheet in (rpw_pre, rpw_post):
    sheet["destination_name"] = sheet["destination_name"].replace(RPW_UNIFY)
rpw_pre = rpw_pre.rename(
    columns={
        "sending location": "access point",
        "coverage": "receiving network coverage",
        "pick-up method": "pickup method",
        "note1": "Standard Note",
    }
)
shared_columns = [column for column in rpw_post.columns if column in rpw_pre.columns]
# duplicates: every shared field except id and date
DUPLICATE_KEY_COLUMNS = [column for column in shared_columns if column not in ("id", "date")]
correction_counts = {"C": 0, "A1": 0, "A2": 0, "B": 0}
sg_correction_counts = {"C": 0, "A1": 0, "A2": 0, "B": 0}


def correct_sheet(sheet):
    numeric = {
        column: pd.to_numeric(sheet[column], errors="coerce")
        for column in [
            "cc1 lcu fee",
            "cc1 lcu amount",
            "cc1 fx margin",
            "cc1 total cost %",
            "cc2 lcu fee",
            "cc2 lcu amount",
            "cc2 fx margin",
            "cc2 total cost %",
        ]
    }
    is_singapore = sheet["source_name"] == "Singapore"
    is_duplicate = sheet.duplicated(subset=DUPLICATE_KEY_COLUMNS)
    # rule A: total must equal fee + margin
    fee_pct_usd200 = numeric["cc1 lcu fee"] / numeric["cc1 lcu amount"] * 100
    mismatch_usd200 = (fee_pct_usd200 + numeric["cc1 fx margin"] - numeric["cc1 total cost %"]).abs() > 0.05
    sheet.loc[mismatch_usd200, "cc1 total cost %"] = (fee_pct_usd200 + numeric["cc1 fx margin"])[
        mismatch_usd200
    ]
    fee_pct_usd500 = numeric["cc2 lcu fee"] / numeric["cc2 lcu amount"] * 100
    mismatch_usd500 = (fee_pct_usd500 + numeric["cc2 fx margin"] - numeric["cc2 total cost %"]).abs() > 0.05
    sheet.loc[mismatch_usd500, "cc2 total cost %"] = (fee_pct_usd500 + numeric["cc2 fx margin"])[
        mismatch_usd500
    ]
    # rule B: margin 15 percentage points above the corridor-quarter median = stale rate
    corridor_quarter_median = (
        numeric["cc1 fx margin"]
        .groupby([sheet["source_name"], sheet["destination_name"], sheet["period"]])
        .transform("median")
    )
    is_stale_rate = (numeric["cc1 fx margin"] - corridor_quarter_median) > 15
    for rule, mask in (
        ("C", is_duplicate),
        ("A1", mismatch_usd200),
        ("A2", mismatch_usd500),
        ("B", is_stale_rate),
    ):
        correction_counts[rule] += int(mask.sum())
        sg_correction_counts[rule] += int((mask & is_singapore).sum())
    return sheet.loc[~(is_duplicate | is_stale_rate)]


rpw_sg_recorded = pd.concat(
    [frame.loc[frame["source_name"] == "Singapore"] for frame in (rpw_pre, rpw_post)], ignore_index=True
)
for column in [
    "cc1 total cost %",
    "cc1 lcu fx rate",
    "cc1 lcu fee",
    "cc1 lcu amount",
    "cc1 fx margin",
    "inter lcu bank fx",
]:
    rpw_sg_recorded[column] = pd.to_numeric(rpw_sg_recorded[column], errors="coerce")
rpw_sg_recorded = (
    rpw_sg_recorded.drop_duplicates().loc[lambda frame: frame["cc1 total cost %"] != 0].reset_index(drop=True)
)
rpw_sg_recorded["year"] = rpw_sg_recorded["period"].str.slice(0, 4).astype(int)
rpw_pre = correct_sheet(rpw_pre).reset_index(drop=True)
rpw_post = correct_sheet(rpw_post).reset_index(drop=True)
wdi_meta = pd.read_excel(WDI_PATH, sheet_name="Metadata - Countries", engine="xlrd")
wdi_country_codes = set(wdi_meta.loc[wdi_meta["Region"].notna(), "Country Code"])
wdi_raw = wdi_raw[wdi_raw["Country Code"].isin(wdi_country_codes)].reset_index(drop=True)
wdi_raw["Country Name"] = wdi_raw["Country Name"].replace(
    {
        "Viet Nam": "Vietnam",
        "Turkiye": "Türkiye",
        "Somalia, Fed. Rep.": "Somalia",
        "Cote d'Ivoire": "Côte d'Ivoire",
    }
)
mci_raw["Country"] = mci_raw["Country"].replace(
    {
        "Egypt": "Egypt, Arab Rep.",
        "Turkey": "Türkiye",
        "Yemen": "Yemen, Rep.",
        "Gambia": "Gambia, The",
        "Congo, Democratic Republic": "Congo, Dem. Rep.",
        "Kyrgyzstan": "Kyrgyz Republic",
        "Laos": "Lao PDR",
        "Cote d'Ivoire": "Côte d'Ivoire",
        "Ivory Coast": "Côte d'Ivoire",
    }
)
for column in ("origin", "destination"):
    riks_raw[column] = (
        riks_raw[column]
        .str.replace("; ", ", ", regex=False)
        .replace({"United States of America": "United States", "Dominican Rep.": "Dominican Republic"})
    )
BRM_RENAMES = {"Turkey": "Türkiye", "Swaziland": "Eswatini", "Cote d'Ivoire": "Côte d'Ivoire"}
brm_raw = brm_raw.rename(index=BRM_RENAMES, columns=BRM_RENAMES)
rpw_all = pd.concat([rpw_pre[shared_columns], rpw_post[shared_columns]], ignore_index=True)
NUMERIC_COLUMNS = [
    "cc1 lcu amount",
    "cc1 lcu fee",
    "cc1 fx margin",
    "cc1 total cost %",
    "cc2 lcu amount",
    "cc2 lcu fee",
    "cc2 fx margin",
    "cc2 total cost %",
    "inter lcu bank fx",
    "cc1 lcu fx rate",
]
rpw_all[NUMERIC_COLUMNS] = rpw_all[NUMERIC_COLUMNS].apply(pd.to_numeric, errors="coerce")
rpw_all["year"] = rpw_all["period"].str.slice(0, 4).astype(int)
rpw_all["quarter"] = rpw_all["period"].str.slice(5, 7)
rpw_all["cc1 fee %"] = rpw_all["cc1 lcu fee"] / rpw_all["cc1 lcu amount"] * 100
rpw_all["transparent"] = rpw_all["transparent"].str.strip().str.lower()
rpw_all["corridor"] = rpw_all["source_name"] + " -> " + rpw_all["destination_name"]
rpw_all = rpw_all.drop_duplicates().loc[lambda frame: frame["cc1 total cost %"] != 0].reset_index(drop=True)
# non-transparent quotes carry a placeholder margin of 0
placeholder_usd200 = rpw_all["transparent"].eq("no") & rpw_all["cc1 fx margin"].eq(0)
placeholder_usd500 = rpw_all["transparent"].eq("no") & rpw_all["cc2 fx margin"].eq(0)
rpw_all.loc[placeholder_usd200, "cc1 fx margin"] = np.nan
rpw_all.loc[placeholder_usd500, "cc2 fx margin"] = np.nan
correction_counts["N"] = int(placeholder_usd200.sum())
sg_correction_counts["N"] = int((placeholder_usd200 & (rpw_all["source_name"] == "Singapore")).sum())
```

### 4.2.2 Name harmonisation and quote-level corrections (all datasets)<a id='sec422'></a>

1. Double-entered quotes that are identical in all 38 analysis fields (215 on Singapore's corridors), differing only in the running id or by a collection date in the same quarter are dropped to keep only one copy. These quotes elude pandas' drop duplicate function due to having different id fields. As quote counts are used as a measure of market size and mean prices or margins are used as a summary of a market, duplicates will corrupt the data.

2. Names are fixed as the datasets use multiple spellings (even within the RPW itself), for example, Turkey/Türkiye, Swaziland/Eswatini, Viet Nam/Vietnam, North Macedonia/Former Yugoslav Republic of Macedonia. Merges are used to combine all data to a single country entry.

3. Quotes where the total cost is computed wrongly are recomputed computationally. For instance, in some quotes, only the margin was included in the total, omitting the fee.  The data breakdown is available to recompute. 249 USD-200 leg totals are recomputed worldwide (72 in Singapore). (Appendix A)

4. 312 stale exchange rates are removed (8 in Singapore). A quote whose FX margin sits highly above its corridor-quarter's median is an outdated rate contradicted by the 6–22 other firms in the same cell and by the same firm's own adjacent quarters, likely obtained by a web scraper. The rule selected was that any quote whose FX margin is more than 15 percentage points above its corridor-quarter's median margin is outdated and needs to be dropped, this test is conducted specific to the corridor's own exchange rate.

5. Non-transparent quotes (i.e. firms that do not disclose their exchange rate) are kept. RPW stores them with an FX margin of exactly 0 rather than as missing, so they understate mean margins instead of dropping out of the averages. If their FX margin is zero, it is replaced with NaN and kept to distinguish them from transparent quotes quoting a margin of zero, and they are not considered when margins are compared and calculated.

These corrections run at the source over both RPW sheets before any table is built.


### 4.2.3 Worldwide scope rule<a id='sec423'></a>

Corridors where the receiving country is in a higher World Bank income band than the sender, or is itself high income (per the WDI metadata of Dataset 5), are dropped throughout (see Introduction). The Netherlands Antilles had missing data for income level; corridors to the Netherlands Antilles were dropped as it was a dissolved country whose successors were high income.

```python
# keep richer-to-poorer corridors only; Netherlands Antilles has no WDI row
INCOME_RANK = {"Low income": 0, "Lower middle income": 1, "Upper middle income": 2, "High income": 3}
wdi_income_rank = (
    wdi_meta.dropna(subset=["IncomeGroup"]).set_index("TableName")["IncomeGroup"].map(INCOME_RANK)
)
WDI_ALIAS = {
    "Turkey": "Türkiye",
    "Vietnam": "Viet Nam",
    "Czech Republic": "Czechia",
    "Macedonia, FYR": "North Macedonia",
    "Swaziland": "Eswatini",
    "Somalia": "Somalia, Fed. Rep.",
}


def in_scope(quotes):
    source_rank = quotes["source_name"].replace(WDI_ALIAS).map(wdi_income_rank)
    destination_rank = quotes["destination_name"].replace(WDI_ALIAS).map(wdi_income_rank)
    to_high_income = (destination_rank == INCOME_RANK["High income"]) | quotes["destination_name"].eq(
        "Netherlands Antilles"
    )
    uphill = source_rank < destination_rank
    return quotes.loc[~to_high_income & ~uphill]
```

**Code Docu:** Defines the helpers used throughout: a pandas Styler for tables, period-to-time conversion, a rolling 12-quarter slope fitted with sklearn LinearRegression and its zero crossing (the breakpoint), hinge fits with numpy lstsq, the tournament summary table and the share stackplot.

```python
# table CSS, inlined (was table_style.json)
TABLE_CSS = [
    {
        "selector": "th.col_heading",
        "props": "text-align:right; font-weight:600; border-bottom:1.5px solid #888; padding:3px 12px;",
    },
    {
        "selector": "th.row_heading",
        "props": "text-align:left; font-weight:400; padding:3px 14px 3px 0; color:#222;",
    },
    {"selector": "td", "props": "text-align:right; padding:3px 12px; border:none;"},
    {"selector": "", "props": "border-collapse:collapse; font-size:12.5px;"},
    {
        "selector": "caption",
        "props": "caption-side:top; text-align:left; color:#555; font-size:12px; padding:0 0 5px 0;",
    },
]


def pretty_table(table, bold_min=(), bold_max=(), fmt="{:.3f}", na="-", caption=None):
    styler = table.style.format(fmt, na_rep=na).set_table_styles(TABLE_CSS)
    for column in bold_min:
        styler = styler.highlight_min(subset=[column], props="font-weight:bold; background-color:#eaf1f8;")
    for column in bold_max:
        styler = styler.highlight_max(subset=[column], props="font-weight:bold; background-color:#eaf1f8;")
    if caption:
        styler = styler.set_caption(caption)
    display(styler)


COMPACT_TABLE_CSS = TABLE_CSS + [
    {"selector": "", "props": "font-size: 8.5px;"},
    {"selector": "td", "props": "padding: 1px 3px;"},
    {
        "selector": "th.col_heading",
        "props": "max-width: 46px; white-space: normal; line-height: 1.1; padding: 1px 3px;",
    },
    {"selector": "th.row_heading", "props": "padding: 1px 6px 1px 0; white-space: nowrap;"},
]


def side_by_side(*tables):
    """Show several tables in one compact row; wraps onto a second row when the page is too narrow."""
    html = "".join(
        (table.style if isinstance(table, pd.DataFrame) else table)
        .format(lambda value: f"{value:.3f}" if isinstance(value, float) else value, na_rep="-")
        .set_table_styles(COMPACT_TABLE_CSS)
        .to_html()
        for table in tables
    )
    display(
        HTML('<div style="display:flex; flex-wrap:wrap; gap:14px; align-items:flex-start">' + html + "</div>")
    )


def smart_eligible(quotes):
    """Services that count towards SmaRT under the World Bank rules: transparent, delivered in 5 days or less,
    and accessible (receiving-network coverage above Low stands in for the geographic-proximity rule)."""
    return quotes[
        (quotes["transparent"] == "yes")
        & (quotes["speed actual"].astype(str).str.strip().str.lower() != "6 days or more")
        & (quotes["receiving network coverage"].astype(str).str.strip().str.lower() != "low")
    ]


def smart_mean(costs):
    """Mean of the three cheapest eligible services; a corridor-quarter with fewer than three has no SmaRT value."""
    return costs.nsmallest(3).mean() if len(costs) >= 3 else np.nan


def period_time(period):
    return int(period[:4]) + (int(period.split("_")[1].rstrip("Q")) - 1) / 4


def qlabel(time):
    return f"{int(time)} Q{min(int((time % 1) * 4) + 1, 4)}"


# breakpoint = where the 12-quarter rolling slope first turns positive
def rolling_slope(series, window=12):
    values = series.dropna()
    windows = [values.iloc[start : start + window] for start in range(len(values) - window + 1)]
    return pd.Series(
        [
            LinearRegression().fit(window.index.values.reshape(-1, 1), window.values).coef_[0]
            for window in windows
        ],
        index=[window.index.values.mean() for window in windows],
    )


SLOPE_WINDOW = 12


def slope_zero_crossing(slopes):
    after = slopes.loc[slopes.idxmin() :]
    for left, right in zip(after.index[:-1], after.index[1:]):
        if after[left] < 0 <= after[right]:
            return left + (right - left) * (-after[left]) / (after[right] - after[left])
    return None


def trend_breakpoint(series, fallback="level"):
    """Quarter where the 12-quarter rolling trend first turns from falling to rising. If it never does:
    fallback="level" -> quarter of the lowest 4-quarter mean; "slope" -> quarter of steepest decline; None -> None.
    """
    slopes = rolling_slope(series.dropna(), SLOPE_WINDOW)
    breakpoint = slope_zero_crossing(slopes)
    if breakpoint is not None or fallback is None:
        return breakpoint
    return slopes.idxmin() if fallback == "slope" else series.rolling(4, min_periods=2).mean().idxmin()


def tournament_summary(breakpoints):
    """One row per series: breakpoint used, hinge + AR(1) holdout RMSE, and the best holdout model against the naive baseline."""
    rows = []
    for label, breakpoint in breakpoints.items():
        table = TOURN_TABLES[label]
        candidates = table.drop(index="naive (last value)")
        best = candidates["holdout RMSE"].idxmin()
        rows.append(
            (
                label,
                qlabel(breakpoint),
                table.loc["hinge + AR(1) errors", "holdout RMSE"],
                best,
                candidates.loc[best, "holdout RMSE"],
                table.loc["naive (last value)", "holdout RMSE"],
                candidates.loc[best, "DM p vs naive"],
            )
        )
    return pd.DataFrame(
        rows,
        columns=[
            "construction",
            "breakpoint",
            "hinge + AR(1) holdout RMSE",
            "best model (holdout)",
            "best holdout RMSE",
            "naive holdout RMSE",
            "DM p vs naive",
        ],
    ).set_index("construction")


def share_stack(ax, fee, margin, title, xlabel=""):
    """100% stacked area of the FX-margin share of the price (4-quarter rolling) with the 50% line; title may use {0}/{1} for the first/last share."""
    fee_smooth, margin_smooth = fee.rolling(4, min_periods=2).mean(), margin.rolling(4, min_periods=2).mean()
    margin_share = (margin_smooth / (fee_smooth + margin_smooth) * 100).dropna()
    ax.stackplot(
        margin_share.index.values,
        100 - margin_share.values,
        margin_share.values,
        alpha=0.9,
        labels=["fee share", "FX margin share"],
    )
    ax.axhline(50, color="black", linestyle=":")
    ax.set(
        ylim=(0, 100),
        yticks=[0, 50, 100],
        xlabel=xlabel,
        title=title.format(margin_share.iloc[:4].mean(), margin_share.iloc[-4:].mean()),
    )
    return margin_share


# piecewise-linear trends, knots searched by least squares
def hinge_design(time, knots):
    return np.column_stack([np.ones_like(time), time] + [np.maximum(0.0, time - knot) for knot in knots])


def hinge_ols(time, values, knots):
    design = hinge_design(time, knots)
    beta, *_ = np.linalg.lstsq(design, values, rcond=None)
    return beta, design @ beta, float(((values - design @ beta) ** 2).sum())


def search_hinge_knots(time, values, n_knots, min_segment=8):
    candidates = [point for point in time if time[min_segment - 1] < point < time[-min_segment]]
    if n_knots == 1:
        options = [[knot] for knot in candidates]
    else:
        options = [
            [first_knot, second_knot]
            for position, first_knot in enumerate(candidates)
            for second_knot in candidates[position + 1 :]
            if second_knot - first_knot >= min_segment * 0.25
        ]
    return min(options, key=lambda knot_set: hinge_ols(time, values, knot_set)[2])
```

**Code Docu:** Builds a corridor-by-quarter table of mean cost with pandas pivot_table, fills the four unsurveyed 2011-12 quarters with the neighbour average, and takes the equal-weight mean across corridors.

```python
world_by_corridor = rpw_world.pivot_table(
    index="period", columns="corridor", values="cc1 total cost %", aggfunc="mean"
).reindex(PERIODS)
world_by_corridor.index = world_by_corridor.index.map(period_time)


# 2011-12 were half-yearly; fill the gaps with neighbour means
def fill_single_gaps(series):
    series = series.dropna().sort_index()
    quarter_grid = np.round(np.arange(series.index.min(), series.index.max() + 1e-9, 0.25), 2)
    series = series.reindex(quarter_grid)
    return series.fillna((series.shift(1) + series.shift(-1)) / 2)


world_corridor_mean = fill_single_gaps(world_by_corridor.mean(axis=1))
```

Within each research question, the worldwide panel (b) is analysed before the Singapore panel (a). The worldwide sample (354 in-scope corridors) is where the formal trend models are fitted, and the Singapore series, the focus of this study, is then read against that global baseline to identify trends, patterns, models, and correlations that might be important for policymakers. 

Multiple metrics are used to aggregate the cost over 215,655 in-scope quotes in 354 corridors at 57 quarterly time intervals (53 surveyed, 4 interpolated in 2011–12). (Appendix A) The World Bank SDG uses the *simple average of all transparent quotes* to calculate the SDG 10.c.1 headline number, the 3% target for the price of remittance in corridors.  [(United Nations Statistics Division, 2026)](#ref18) However, this can be unrepresentative as a corridor with a larger number of quotes is given higher weight, and the surveyed number of quotes may not be fully representative of demand in the corridors. Hence, two other metrics are prioritized in this analysis. The flow-weighted average, also used by the World Bank but more peripherally  [(World Bank, 2026)](#ref28), where the average cost in each corridor is aggregated and then corridors are weighted by amount of remittance flow (2018 BRM, Dataset 2), and one where all corridors are weighted equally, created for this report, are also explored to observe if different nuances can lead to relevant conclusions. 

The SmaRT-style average is used to determine if any corridors fall above the 5% ceiling, taking the three cheapest transparent services per corridor [(United Nations Statistics Division, 2026)](#ref18)[(World Bank, n.d.-d)](#ref23), keeping only services delivered in five days or less with receiving-network coverage above Low, and only where a corridor has at least three such services in the quarter, following the SmaRT rules. However, the cheapest services may not be representative of what a migrant pays as they may be bank account transfers requiring high internet connectivity, when a migrant's community has no internet access and requires cash pickup. Hence, the mean remittance price is also plotted, to draw separate conclusions. 

A similar analysis of Singapore's corridors hence follows.

```python
#  fence so one outlier corridor does not set the line
def highest_non_outlier(panel):
    lower_quartile, upper_quartile = panel.quantile(0.25, axis=1), panel.quantile(0.75, axis=1)
    inliers = panel.where(panel.le(upper_quartile + 1.5 * (upper_quartile - lower_quartile), axis=0)).max(
        axis=1
    )
    return fill_single_gaps(inliers).rolling(4, min_periods=2).mean()


def build_smart_panel(quotes, key):
    panel = smart_eligible(quotes).groupby(["period", key])["cc1 total cost %"].apply(smart_mean).unstack()
    panel.index = panel.index.map(period_time)
    return panel.sort_index()


smart_by_corridor = build_smart_panel(world_transparent_quotes, "corridor")
sg_smart_by_corridor = build_smart_panel(sg_transparent_quotes, "destination_name")
scope_panels = [
    ("worldwide (354 corridors)", world_by_corridor, smart_by_corridor),
    ("Singapore (nine corridors)", sg_by_corridor, sg_smart_by_corridor),
]
fence_series = [
    pd.DataFrame(
        {
            "highest non-outlier corridor (mean quote)": highest_non_outlier(mean_costs),
            "highest non-outlier corridor (SmaRT-style)": highest_non_outlier(smart_costs),
        }
    )
    for _, mean_costs, smart_costs in scope_panels
]
y_top = (
    max(float(frame.iloc[:, 0].max()) for frame in fence_series) * 1.12
)  # one shared y-axis for both panels
fig, axes = plt.subplots(1, 2, figsize=(12, 3.9), sharey=True)
for ax, (scope_name, _, _), fence_frame in zip(axes, scope_panels, fence_series):
    fence_frame.plot(ax=ax, color=["black", "C3"], title=scope_name, legend=False, xlabel="")
    ax.axhline(5, color="red")
    ax.axhspan(5, y_top, color="red", alpha=0.06)
    ax.set_ylim(0, y_top)
    ax.text(2011.2, 5.2, "5% SDG 10.c corridor ceiling", color="red", va="bottom")
axes[0].set_ylabel("total cost of sending USD 200\n(% of amount, 4-qtr rolling)")
fig.legend(*axes[0].get_legend_handles_labels(), loc="lower center", ncol=2, bbox_to_anchor=(0.5, -0.06))
plt.tight_layout()
plt.show()
```

### b. Modelling and forecastability

#### b1. Model Introductions

Noting that there is a breakpoint at 2021 Q3 as the previous graphs have explained, a quadratic trend, hinged trend (two linear trends with a breakpoint), Holt Linear Trend, a time-series forecasting technique that extends simple exponential smoothing (decreasing weights to older data) to handle data with a trend, Holt Damped Trend, a variation of the former optimized for a plateau, and ARIMA models were used.

The naive model is a benchmark where the extrapolation is solely based on the previous data point;beating naive is necessary for the model to be even considered as an accurate forecast (but not sufficient), hence a preliminary test is to run the models against naive.

The parameters obtained by data obtained via the Holt and ARIMA models provided some insights on the data. The Holt's Damped Trend, when calculations were optimized, set a *$\beta\$* value of zero. This indicated that the model was merely a naive estimate of the data (taking the previous data point) modified to converge and plateau at a certain level.

ARIMA models data by modelling the differences along with regressing the series on its own past values (its most recent *p* lags) and using a moving average of mistakes to quantify the past "shocks" (abrupt changes in the data), serving as a relatively robust model when analysing data with significant fluctuations.

The models are compared with multiple statistical tests to determine which is a useful projection, if any.

The in-sample root-mean-squared-error is the metric taught in the course to determine how closely the model approximates the
data it was trained on, where the individual errors for each point are squared.

The remaining three columns of the evaluation table are read as follows. **resid ac1** is the lag-1 autocorrelation of the model's residuals: if ac1 is close to 1, there is a correlation in the residuals and hence the model is not a good representation as the model failed to capture the carry-over of errors from quarter to quarter. **Ljung-Box p** turns this into a formal test across the first eight lags, with the null hypothesis that the residuals are independent: a p-value below 0.05 rejects that null and marks the model as statistically invalid, however accurate its projections seem, while a high p-value means the model passes. **holdout RMSE** is the out-of-sample test: every model is refitted using only data before 2023 Q2 and made to forecast the final eight quarters it has never seen, and this column records the root-mean-squared-error of those forecasts, which is similar to the testing methods used in the notes.

To draw a conclusion from the table: first discard any model that fails the residual checks as it is statistically invalid. Second, among the surviving models, compare each holdout RMSE against the naive row. Third, prefer the lowest holdout RMSE among the valid models, treating in-sample RMSE only as a consistency check rather than a criterion when attempting to quantify the forecasting value of the model. Additionally, models can be compared to versions trained on data excluding linear interpolated points for maximum accuracy.

DM p vs naive is the Diebold-Mariano test of equal predictive accuracy. A lower holdout RMSE only says a model was better on average over eight quarters; this test asks whether that difference is greater than the naive estimate, by comparing the two models' squared errors quarter by quarter and testing whether their mean difference is distinguishable from zero, with the Harvey-Leybourne-Newbold correction for the small sample, this test is used to reward consistency in accurate, or better than baseline, predictions.

**Code Docu:** For each corridor, fits constant, linear and one- or two-knot hinge trends with numpy lstsq, picks one by BIC, checks residuals with statsmodels acorr_ljungbox and refits with AR(1) errors (SARIMAX) where they fail, then runs the corridor's own tournament and tabulates descriptive and predictive validity.

```python
# per corridor: constant / linear / hinge by BIC, AR(1) refit if Ljung-Box fails
corridor_fits, corridor_fit_rows = {}, []
for dest in sorted(sg_by_corridor.columns):
    series = sg_by_corridor[dest].dropna()
    time, cost, n_obs = series.index.values, series.values, len(series)
    sse_constant = float(((cost - cost.mean()) ** 2).sum())
    one_knot, two_knots = search_hinge_knots(time, cost, 1), search_hinge_knots(time, cost, 2)
    candidates = {
        "constant": ([], sse_constant, 1),
        "linear": ([], hinge_ols(time, cost, [])[2], 2),
        "hinge-1": (one_knot, hinge_ols(time, cost, one_knot)[2], 4),
        "hinge-2": (two_knots, hinge_ols(time, cost, two_knots)[2], 6),
    }
    bic = lambda sse, n_params: n_obs * np.log(max(sse, 1e-9) / n_obs) + n_params * np.log(n_obs)
    name = min(candidates, key=lambda model: bic(candidates[model][1], candidates[model][2]))
    knots, sse, _ = candidates[name]
    beta, pred = (
        (np.array([cost.mean()]), np.full(n_obs, cost.mean()))
        if name == "constant"
        else hinge_ols(time, cost, knots)[:2]
    )
    residuals = cost - pred
    lag1_autocorr = np.corrcoef(residuals[:-1], residuals[1:])[0, 1]
    ljung_box_p = acorr_ljungbox(residuals, lags=[8], return_df=True)["lb_pvalue"].iloc[0]
    spec, trend, coefficients, r2 = name, pred, list(beta[1:]), 1 - sse / max(sse_constant, 1e-12)
    if name != "constant" and ljung_box_p < 0.10:
        design = hinge_design(time - time[0], [knot - time[0] for knot in knots])[:, 1:]
        with warnings.catch_warnings(record=True) as caught_warnings:
            warnings.simplefilter("always")
            ar1_model = SARIMAX(cost, exog=design, order=(1, 0, 0), trend="c").fit(disp=False)
            converged = not any("converge" in str(warning.message).lower() for warning in caught_warnings)
        ljung_box_ar1 = acorr_ljungbox((cost - ar1_model.fittedvalues)[1:], lags=[8], return_df=True)[
            "lb_pvalue"
        ].iloc[0]
        if converged and ljung_box_ar1 >= 0.05:
            params = dict(zip(ar1_model.param_names, ar1_model.params))
            coefficients = [
                params[param_name] for param_name in ar1_model.param_names if param_name.startswith("x")
            ]
            spec, ljung_box_p, lag1_autocorr = name + " + AR(1)", ljung_box_ar1, params["ar.L1"]
    slopes = list(np.cumsum(coefficients)) if name != "constant" else []
    corridor_fits[dest] = (series, trend, knots, spec)
    corridor_fit_rows.append((dest, spec, knots, slopes, r2, lag1_autocorr, ljung_box_p))
CORRIDOR_SUMMARY_ROWS = []
for dest in sorted(sg_by_corridor.columns):
    series = sg_by_corridor[dest].dropna()
    model_tournament(
        series, trend_breakpoint(series, fallback="slope"), f"Singapore corridor {dest}", show_table=False
    )
    table = TOURN_TABLES[f"Singapore corridor {dest}"]
    naive_rmse = table.loc["naive (last value)", "holdout RMSE"]
    valid_models = table.drop(index="naive (last value)").query("`Ljung-Box p` >= 0.05")
    chosen_spec, chosen_lb_p = next((row[1], row[6]) for row in corridor_fit_rows if row[0] == dest)
    recent_level = float(series.iloc[-8:].mean())
    if len(valid_models):
        best_model = valid_models["holdout RMSE"].idxmin()
        CORRIDOR_SUMMARY_ROWS.append(
            (
                dest,
                chosen_spec,
                chosen_lb_p,
                best_model,
                float(valid_models.loc[best_model, "holdout RMSE"]),
                float(naive_rmse),
                float(valid_models.loc[best_model, "holdout RMSE"]) / recent_level * 100,
                float(valid_models.loc[best_model, "DM p vs naive"]),
            )
        )
    else:
        CORRIDOR_SUMMARY_ROWS.append(
            (dest, chosen_spec, chosen_lb_p, "none valid", np.nan, float(naive_rmse), np.nan, np.nan)
        )
corridor_summary = (
    pd.DataFrame(
        CORRIDOR_SUMMARY_ROWS,
        columns=[
            "corridor",
            "best specification",
            "Ljung-Box p",
            "best valid forecaster",
            "holdout RMSE",
            "naive RMSE",
            "RMSE % of level",
            "DM p vs naive",
        ],
    )
    .set_index("corridor")
    .round(3)
)
pretty_table(corridor_summary, fmt=lambda value: value if isinstance(value, str) else f"{value:.3f}")
```

**Code Docu:** Fits each component series as a hinge or linear trend with AR(1) errors using statsmodels SARIMAX and reads the slopes and p-values.

```python
AR1_RESULTS = {}


def fit_ar1_trend(series, knots, label):
    series = series.dropna()
    time, values = series.index.values, series.values
    time_centred = time - time[0]
    design = hinge_design(time_centred, [knot - time[0] for knot in knots])[:, 1:]
    ols_beta = hinge_ols(time, values, knots)[0]
    model = SARIMAX(values, exog=design, order=(1, 0, 0), trend="c").fit(disp=False)
    params_by_name = dict(zip(model.param_names, model.params))
    phi = [value for param_name, value in params_by_name.items() if param_name.startswith("ar.")][0]
    params, pvalues, covariance = (
        np.asarray(model.params),
        np.asarray(model.pvalues),
        np.asarray(model.cov_params()),
    )
    slope_positions = [
        position for position, param_name in enumerate(model.param_names) if param_name.startswith("x")
    ]
    result = {
        "phi": phi,
        "r2_ols": 1 - hinge_ols(time, values, knots)[2] / float(((values - values.mean()) ** 2).sum()),
    }
    if knots:
        before_idx, bend_idx = slope_positions[0], slope_positions[1]
        slope_after = params[before_idx] + params[bend_idx]
        slope_after_se = np.sqrt(
            covariance[before_idx, before_idx]
            + covariance[bend_idx, bend_idx]
            + 2 * covariance[before_idx, bend_idx]
        )
        result.update(
            ols_before=ols_beta[1],
            ols_after=ols_beta[1] + ols_beta[2],
            ar_before=params[before_idx],
            ar_before_p=pvalues[before_idx],
            ar_after=slope_after,
            ar_after_p=2 * (1 - stats.norm.cdf(abs(slope_after / slope_after_se))),
        )
    else:
        result.update(
            ols_slope=ols_beta[1], ar_slope=params[slope_positions[0]], ar_slope_p=pvalues[slope_positions[0]]
        )
    AR1_RESULTS[label] = result


fit_ar1_trend(world_fee, [world_fee_breakpoint], "fee (hinge)")
fit_ar1_trend(world_margin, [], "FX margin (linear)")
fit_ar1_trend(world_margin_share, [], "margin share (linear)")
fit_ar1_trend(sg_fee, [sg_fee_breakpoint], "SG fee (hinge)")
fit_ar1_trend(sg_margin, [], "SG FX margin (linear)")
fit_ar1_trend(sg_margin / (sg_fee + sg_margin) * 100, [], "SG margin share (linear)")
```

#### Data Modelling 

**Code Docu:** Splits the corridors 80/20 once with a seeded numpy permutation, ranks the candidate factors by |Spearman ρ| against cost on the training corridors only and keeps the top five, fits one statsmodels OLS per selected factor on the training corridors with HC3 errors, scores its predictions on the held-out corridors, and draws seaborn regplot and residplot panels on a GridSpec with a summary table.

```python
# Singapore's own value of each corridor factor (used for the Singapore checks printed in Appendix A)
sg_latest_quotes = rpw_sg[rpw_sg["period"].map(period_time).isin(LATEST_QUARTERS)].assign(
    is_bank=lambda data: data["firm_type"].astype(str).str.lower().str.contains("bank"),
    speed_days=lambda data: data["speed actual"].astype(str).str.strip().str.lower().map(SPEED_DAYS),
)
sg_by_destination = sg_latest_quotes.groupby("destination_name")
sg_demand = rq3_sg.set_index("country")
VOLATILITY_COL = "log10 currency\nvolatility (%/qtr)"
sg_factor_values = {
    "bank share\nof quotes (%)": sg_by_destination["is_bank"].mean() * 100,
    "speed\n(days)": sg_by_destination["speed_days"].mean(),
    **{
        label: np.log10(sg_demand[column].where(sg_demand[column] > 0))
        for label, column in [
            ("log10\noutflow (BRM)", "remittance_outflow_2018_usd_millions"),
            ("log10\nflow", "migrant flow"),
            ("log10\nstock", "migrant stock"),
        ]
    },
    VOLATILITY_COL: corridor_factors.loc[
        corridor_factors.index.str.startswith("Singapore -> "), VOLATILITY_COL
    ].rename(lambda name: name.split(" -> ")[1]),
}
COST_COL, FEE_COL, MARGIN_COL = COST_COMPONENTS
Y_LIMITS = {COST_COL: (0, 25), FEE_COL: (0, 25), MARGIN_COL: (-4, 15)}

# ---- Train/test split: made once, BEFORE any predictor is chosen ----
# 20% of the priced corridors are held out. The predictors are then ranked on the training corridors only,
# every model is fitted on the training corridors only, and the held-out corridors are used for nothing
# except scoring the predictions.
TEST_FRACTION, SPLIT_SEED, N_SELECTED = 0.2, 4132, 5
priced_corridors = corridor_factors.dropna(subset=[COST_COL])
shuffled_corridors = np.random.default_rng(SPLIT_SEED).permutation(priced_corridors.index.to_numpy())
n_test_corridors = int(len(shuffled_corridors) * TEST_FRACTION)
test_factors = priced_corridors.loc[shuffled_corridors[:n_test_corridors]]
train_factors = priced_corridors.loc[shuffled_corridors[n_test_corridors:]]
assert not set(train_factors.index) & set(test_factors.index)

# ---- Predictor selection, on the training corridors only ----
candidate_factors = [column for column in corridor_factors.columns if column not in COST_COMPONENTS]


def train_spearman(factor):
    pair = train_factors[[factor, COST_COL]].dropna()
    return stats.spearmanr(pair[factor], pair[COST_COL])[0]


train_rho = pd.Series({factor: train_spearman(factor) for factor in candidate_factors})
selected_factors = list(train_rho.abs().sort_values(ascending=False).index[:N_SELECTED])


def fit_factor_model(factor, component):
    """OLS of one cost component on one factor: fitted on the training corridors (HC3 errors),
    then scored on its predictions for the held-out corridors."""
    train = train_factors[[factor, component]].dropna()
    test = test_factors[[factor, component]].dropna()
    model = sm.OLS(train[component], sm.add_constant(train[factor].astype(float))).fit(cov_type="HC3")
    predicted = model.predict(sm.add_constant(test[factor].astype(float), has_constant="add"))
    test_mse = ((test[component] - predicted) ** 2).mean()
    naive_mse = (
        (test[component] - train[component].mean()) ** 2
    ).mean()  # baseline: predict the training mean
    scores = dict(rmse=test_mse**0.5, rmse0=naive_mse**0.5, tr2=1 - test_mse / naive_mse, test=test)
    return train, model, scores


slope_label = lambda model: f"{model.params.iloc[1]:+.2f}" + ("*" if model.pvalues.iloc[1] < 0.05 else "")
equation_label = lambda model: f"y = {model.params.iloc[0]:.2f} {model.params.iloc[1]:+.2f}x"
TEST_R2_BANDS = [
    (0.26, "strong ≥ 0.26", "#7f2704", "white"),
    (0.13, "moderate 0.13–0.26", "#f16913", "white"),
    (0.02, "weak 0.02–0.13", "#fdd0a2", "black"),
    (0.0, "negligible 0–0.02", "white", "grey"),
    (-9.0, "fails ≤ 0", "white", "grey"),
]

# ---- One model per selected factor: cost and fee panels, each with its residual plot ----
factor_rows = []
fig = plt.figure(figsize=(14, 11))
grid = fig.add_gridspec(N_SELECTED, 4)
for row_index, factor in enumerate(selected_factors):
    panel_pairs = [
        (fig.add_subplot(grid[row_index, 0]), fig.add_subplot(grid[row_index, 1])),
        (fig.add_subplot(grid[row_index, 2]), fig.add_subplot(grid[row_index, 3])),
    ]
    factor_label = factor.replace("\n", " ")
    fits = {component: fit_factor_model(factor, component) for component in COST_COMPONENTS}
    cost_train, cost_model, cost_test = fits[COST_COL]
    fee_model, margin_model = fits[FEE_COL][1], fits[MARGIN_COL][1]
    factor_rows.append(
        (
            factor_label,
            train_rho[factor],
            len(cost_train),
            len(cost_test["test"]),
            cost_model.params.iloc[0],
            cost_model.params.iloc[1],
            cost_model.pvalues.iloc[1],
            cost_model.rsquared,
            cost_test["tr2"],
            cost_test["rmse"],
            cost_test["rmse0"],
            slope_label(fee_model),
            slope_label(margin_model),
        )
    )
    for (ax, ax_resid), component in zip(panel_pairs, [COST_COL, FEE_COL]):
        train, model, test_scores = fits[component]
        sns.regplot(
            data=train,
            x=factor,
            y=component,
            ci=None,
            ax=ax,
            scatter_kws=dict(s=9, alpha=0.35),
            line_kws=dict(color="red", label="line fitted on the training corridors"),
            label="training corridors (80%)",
        )
        ax.scatter(
            test_scores["test"][factor],
            test_scores["test"][component],
            s=14,
            facecolors="none",
            edgecolors="C1",
            linewidths=0.9,
            label="held-out corridors (20%)",
        )
        ax.set(xlabel=factor_label, ylabel=component.replace("\n", " "), ylim=Y_LIMITS[component])
        ax.set_title(
            f"{component.split(chr(10))[0]}: {equation_label(model)} (pp), p {model.pvalues.iloc[1]:.0e}",
            fontsize=8,
        )
        tag_axes(ax, f"train R² {format_3sf(model.rsquared)}", 0.02, "left", model.rsquared, R2_BANDS)
        tag_axes(
            ax,
            f"test R² {format_3sf(test_scores['tr2'], True)}",
            0.98,
            "right",
            test_scores["tr2"],
            TEST_R2_BANDS,
            use_abs=False,
        )
        sns.residplot(data=train, x=factor, y=component, ax=ax_resid, scatter_kws=dict(s=9, alpha=0.35))
        ax_resid.set(xlabel=factor_label, ylabel="residual (pp)")
        ax_resid.set_title(f"{component.split(chr(10))[0]} residuals (training corridors)", fontsize=8)
fig.legend(
    *panel_pairs[0][0].get_legend_handles_labels(), loc="lower center", ncol=3, bbox_to_anchor=(0.5, -0.01)
)
fig.suptitle(
    "One model per factor, worldwide: fitted on the training corridors, scored on the held-out corridors"
)
fig.tight_layout(rect=[0, 0.015, 1, 0.985])
plt.show()
pretty_table(
    pd.DataFrame(
        factor_rows,
        columns=[
            "factor",
            "train ρ",
            "n train",
            "n test",
            "intercept",
            "slope on cost",
            "p (HC3)",
            "train R²",
            "test R²",
            "test RMSE",
            "naive RMSE",
            "slope on fee",
            "slope on margin",
        ],
    ).set_index("factor"),
    fmt={
        "train ρ": "{:+.2f}",
        "n train": "{:d}",
        "n test": "{:d}",
        "intercept": "{:.2f}",
        "slope on cost": "{:+.2f}",
        "p (HC3)": "{:.1e}",
        "train R²": "{:.2f}",
        "test R²": "{:+.2f}",
        "test RMSE": "{:.2f}",
        "naive RMSE": "{:.2f}",
    },
)
```

As the fee is the most affected portion of the data, while the FX margin regressions are usually flat or with mild gradients, the data is modelled for the cost and fee.

To keep the test honest, the corridors are split 80/20 before any predictor is chosen: the five factors are selected, and every line is fitted, on the 261 training corridors only, and the 65 held-out corridors (hollow orange points) are used only to score the predictions. [Figure 11](#Figure-11) remains a descriptive view of all corridors; the same five factors lead on the training corridors alone. (Appendix A)

Out of sample, bank share of quotes is the only single factor that is strongly predictive (test R^2 0.31). Sender bank share and stock are moderately predictive (both 0.16) and speed weakly so (0.11), while outflow on its own does not beat the training-mean baseline on the held-out corridors (test R^2 −0.01), so its slope is read as an association rather than as a predictor.

The percentage of quotes in the corridor sent by banks, the bank share of quotes has the largest correlation with the cost, with a moderate significance R^2 and strong test R^2. When corridors are more bank-dominated, they are more expensive, as banks have more secure, regulated, and integrated networks and bank quotes are more expensive. The entry of Online Money Transfer Operators (MTOs) and Fintech Peer-to-Peer (P2P) platforms (and other alternatives to banks) with lower prices also induce prices in a corridor to decrease due to competition causing even banks to lower prices, and hence the more bank-dominated a corridor is, it will be more expensive. 

The same conclusion applies for the sender bank share, which instead measures how bank-heavy the sending country's whole remittance industry is, and not just the corridor, regardless of destination. 

Outflow and stock, entering as log₁₀ values so their slopes read as percentage points per tenfold increase, both have moderate-to-weak negative correlations with the components of cost. The correlation can be hypothesized to be due to economies of scale, where when more money flows (outflow) or the diaspora is larger (stock), this is indicative of a larger market where economies of scale would pull costs down.

**Code Docu:** Fits a joint statsmodels OLS of cost on the five selected factors with HC3 errors on the training corridors, computes VIFs, predicts the held-out corridors to get the test R-squared, and plots actual against predicted and the residuals for both sets with a summary table.

```python
# All selected factors at once, on the same split: fitted on the training corridors, predicting the held-out ones.
joint_columns = selected_factors + [COST_COL]
joint_train = train_factors[joint_columns].dropna()
joint_test = test_factors[joint_columns].dropna()
joint_design = sm.add_constant(joint_train[selected_factors].astype(float))
joint_model = sm.OLS(joint_train[COST_COL], joint_design).fit(cov_type="HC3")
joint_test_pred = joint_model.predict(
    sm.add_constant(joint_test[selected_factors].astype(float), has_constant="add")
)
joint_test_mse = ((joint_test[COST_COL] - joint_test_pred) ** 2).mean()
joint_naive_mse = ((joint_test[COST_COL] - joint_train[COST_COL].mean()) ** 2).mean()
joint_test_r2 = 1 - joint_test_mse / joint_naive_mse

# standardised betas compare the factors on one scale; VIF checks collinearity between them
joint_std = joint_train.std()
joint_rows = [
    (
        factor.replace("\n", " "),
        joint_model.params[factor],
        joint_model.params[factor] * joint_std[factor] / joint_std[COST_COL],
        joint_model.pvalues[factor],
        variance_inflation_factor(joint_design.values, position + 1),
    )
    for position, factor in enumerate(selected_factors)
]

fig, (ax_predicted, ax_residual) = plt.subplots(1, 2, figsize=(12, 4.6))
ax_predicted.scatter(
    joint_model.fittedvalues,
    joint_train[COST_COL],
    s=10,
    alpha=0.25,
    label=f"training corridors (n {len(joint_train)})",
)
ax_predicted.scatter(
    joint_test_pred,
    joint_test[COST_COL],
    s=22,
    color="C1",
    edgecolors="black",
    linewidths=0.4,
    label=f"held-out corridors (n {len(joint_test)})",
)
diagonal = [0, float(max(joint_train[COST_COL].max(), joint_test[COST_COL].max())) * 1.02]
ax_predicted.plot(diagonal, diagonal, color="grey", ls=":")
ax_predicted.set(xlabel="predicted cost (%)", ylabel="actual cost (%)", title="(a) actual vs predicted")
ax_predicted.legend(loc="lower right")
tag_axes(
    ax_predicted,
    f"train R² {format_3sf(joint_model.rsquared)}",
    0.02,
    "left",
    joint_model.rsquared,
    R2_BANDS,
)
tag_axes(
    ax_predicted,
    f"test R² {format_3sf(joint_test_r2, True)} on the held-out 20%",
    0.98,
    "right",
    joint_test_r2,
    TEST_R2_BANDS,
    use_abs=False,
)
ax_residual.scatter(joint_model.fittedvalues, joint_model.resid, s=10, alpha=0.25)
ax_residual.scatter(
    joint_test_pred,
    joint_test[COST_COL] - joint_test_pred,
    s=22,
    color="C1",
    edgecolors="black",
    linewidths=0.4,
)
ax_residual.axhline(0, color="grey", ls=":")
ax_residual.set(xlabel="predicted cost (%)", ylabel="residual (pp)", title="(b) residuals vs predicted")
plt.tight_layout()
plt.show()
pretty_table(
    pd.DataFrame(
        joint_rows, columns=["factor", "coef (percentage points)", "std beta", "p (HC3)", "VIF"]
    ).set_index("factor"),
    fmt={"coef (percentage points)": "{:+.2f}", "std beta": "{:+.2f}", "p (HC3)": "{:.1e}", "VIF": "{:.1f}"},
)
```

```python
# Spearman rho and p-value of every Figure 11 factor against total cost, plus the five-factor joint model of Figure 14 as a last row: |rho| banded by Cohen's (1988) conventions as in Figure 11, p banded by the conventional 0.05 / 0.01 / 0.001 significance levels.
P_BANDS = [
    (0.0, "p < 0.001", "#08306b", "white"),
    (0.001, "p < 0.01", "#2171b5", "white"),
    (0.01, "p < 0.05", "#9ecae1", "black"),
    (0.05, "n.s. (p ≥ 0.05)", "white", "grey"),
]
band_label = lambda value, bands: next(label for threshold, label, _, _ in bands if abs(value) >= threshold)
p_band_label = lambda value: next(label for threshold, label, _, _ in reversed(P_BANDS) if value >= threshold)
band_colours = {label: (face, ink) for bands in (RHO_BANDS, P_BANDS) for _, label, face, ink in bands}
factor_table = pd.DataFrame(
    [
        {
            "factor": factor.replace("\n", " "),
            **dict(
                zip(
                    ["rho vs cost", "p"],
                    stats.spearmanr(*corridor_factors[[factor, COST_COMPONENTS[0]]].dropna().values.T),
                )
            ),
        }
        for factor in factors_ranked
    ]
).set_index("factor")
factor_table["rho band"] = factor_table["rho vs cost"].map(lambda value: band_label(value, RHO_BANDS))
factor_table = factor_table.sort_values("rho vs cost", key=abs, ascending=False)
factor_table.loc["five factors jointly (MLR, Figure 14)"] = [
    np.nan,
    joint_model.f_pvalue,
    f"R² {joint_model.rsquared:.2f}, test R² {joint_test_r2:+.2f}",
]
factor_table["significance"] = factor_table["p"].map(p_band_label)
factor_table = factor_table[["rho vs cost", "rho band", "p", "significance"]]
cell_style = lambda label: "background-color:{}; color:{}".format(
    *band_colours.get(label, ("#f2f2f2", "#222"))
)
display(
    factor_table.style.format(
        {"rho vs cost": "{:+.2f}", "p": lambda value: "< 0.001" if value < 0.001 else f"{value:.3f}"},
        na_rep="–",
    )
    .apply(lambda row: [cell_style(row["rho band"])] * 2 + [cell_style(row["significance"])] * 2, axis=1)
    .set_table_styles(
        TABLE_CSS
        + [
            {"selector": "", "props": "font-size: 9px;"},
            {"selector": "th, td", "props": "padding: 1px 6px; white-space: nowrap;"},
        ]
    )
)
```

Once bank share is held constant in the model (MLR with all five significant factors), the speed coefficient flips sign (−1.54 percentage points a day), so there is no price of urgency left to explain; a corridor is not expensive because it is slow, it is slow and expensive because it is served by banks ([Figure 14](#Figure-14)) (Dataset 1).

# 6. Limitations & Assumptions<a id='limitations'></a>

## 6.1 Limitations<a id='sec61'></a>

### Research Question 1 — the cost trend and the SDG targets

The SmaRT figure applies the World Bank's transparency, five-day and three-service rules exactly, but its accessibility rule cannot be reconstructed, as RPW does not include data on ease of access such as branch proximity or account/credit requirements, so services with a Low receiving-network coverage rating are excluded as a proxy. Most quotes in RPW are likely well accessible (or else, they may not have made it into the study), so this is likely not significant.

Data is more populated in the RPW datasets in the more recent time period, Sri Lanka was first surveyed in 2015 Q4, and Pakistan had less than five quotes for each quarter from 2011 to 2016, hence data from the first half of the survey may not have been so reliable. However, this is not a key concern as the important conclusions come from recent quotes and the areas fundamental to analysis and projection have reliable data.

### Research Question 2 — fee and FX margin

FX margin is only observable where the firm discloses its rate. Quotes flagged non-transparent, likely where data was not provided to RPW data collectors/scrapers when they pretended to purchase data, carry no usable margin, so the fee/margin split is measured on a self-selected subset of firms, leading to bias towards the data likely representing a lower set of the prices on offer. However, these only comprise 17 quotes in Singapore's corridors and are not crucial to the primary analysis.

### Research Question 3 — corridor factors

RIKS covers all 9 corridors annually, but its values are 'model-imputed' *(sic)*, not observed. The RIKS data reports more Pakistanis than Bangladeshis (127,840 vs 80,658 in 2020), and roughly 8x more Pakistanis than Filipinos, contrary to the expected values from qualitative common knowledge. (Appendix A) Migrant stock is a proxy for remittance demand, not a measure of it. It does not represent how much money is actually remitted. Hence, it is combined with the bilateral remittance matrix to provide a second scope or angle for data analytics, and both concur.

Every RQ3 factor is a single-period snapshot of the most recent available data (RIKS stock 2020, bilateral matrix 2018, MCI 2023 to 2025, HDI 2023, inflation 2022 to 2025) set against the latest four-quarter cost window, so the models compare corridors with each other at a rough snapshot in time and cannot track a corridor over time, which is acceptable as RQ3 primarily focuses on current over past data. The most significant issue is that the 2018 matrix also predates the pandemic and covers 337 of the 354 corridors, but it is the most representative as it is the most recent edition of this withdrawn release that has not been pulled. (Appendix A)

Nine corridors are too few to rank corridor-level factors within Singapore, hence Singaporean corridors can only be robustly analysed at the quote level in a few cases causing the analysis to be slightly more limited than worldwide. We will try to analyse Singaporean remittances in the worldwide context to determine policy recommendations for those too. Any further data collection is a future work that can only be resolved by organizations like the World Bank ([Section 7](#7.-Recommendations-or-Further-Works)).


## 6.2 Assumptions<a id='sec62'></a>

For the overall dataset, all corridors where the receiving country is in a higher World Bank income band than the sender, or is itself high income, are dropped entirely; corridors between countries of the same income band below high income are retained. This assumes that these corridors have a minimal expatriate salary portion considered in the data, and thus that these fit entirely with the geographical definition of remittances ([Section 1.4](#1.4-Project-Scope)). It will not be a hundred-percent fit, but it is close enough to render this a reliable study.

USD 200 is taken as the representative transfer size, as it is the SDG 10.c benchmark amount; costs at other sizes (RPW also records a USD 500 leg) were not analysed.


# 7. Recommendations or Further Works<a id='recommendations'></a>
## 7.1 Policy Recommendations 

These recommendations are scoped for the Singaporean government. They follow the associations found in section 5, not tested as causes. 

The government must continue to work towards having all corridors from Singapore meet the 3% mean and try to stop the plateau. 

As bank corridors are dearer, the Singaporean government can provide subsidies for other firms like money-transfer operators or mobile operators, as well as easier regulatory approval, lower licensing costs, and compliance costs under the Payment Services Act [(Monetary Authority of Singapore, 2025)](#ref30). These entities offer quicker, more lightly licensed and, on average, cheaper services [(World Bank, 2026)](#ref28). Hence, by providing subsidies for this type of enterprise, more will be enticed to set up this type of enterprise and offer cheaper prices as operating costs are cheaper and more will enter the market, causing competition to increase and the fee decline to resume until Singaporean corridors meet the 3% mean target [(Australian Competition and Consumer Commission, 2019)](#ref37). Publishing data of which providers serve each corridor can also ease market diversification by making thin corridors more visible to potential entrants [(Committee on Payment and Settlement Systems & World Bank, 2007)](#ref33). Further, remittance agencies can be provided additional subsidies if they handle more transactions within corridors that have higher costs [(Business Recorder, 2026)](#ref40).

As the FX margin is the portion not subjected to the decreasing trends, the government can also strictly mandate that in all remittances from Singapore the FX margin be shown visibly to the consumer so it would be subjected to the same consumer awareness and hence competitive forces that drove the fee down in the first place ([Regulation (EU) 2019/518, 2019](#ref34); [12 C.F.R. § 1005.31, 2026](#ref36)). Concurrently, the practice of advertising 'zero-fee' remittance without considering the costs paid by way of FX margin can be wholly banned through mandating adherence to the optional Singapore FinTech Association's Payments Industry Code of Conduct. ([Singapore FinTech Association, 2026](#ref29); [Consumer Financial Protection Bureau, 2024](#ref35)).

Through MOM's channels to work-permit holders, the cheapest quotes can be advertised, both allowing workers to pay less for remittance and providing an incentive for lower costs in the form of greater publicity for remittance firms [(Ministry of Manpower, 2025)](#ref38).

More data surrounding remittances can be made publicly available/available to academics to inform more country-specific policy-oriented projects like this one. Closely tracking demand in Singapore's corridors, as the bilateral remittance matrix did, will be useful to inform more targeted decisions. An RPW-style mystery-shopping survey for corridors RPW doesn't price (Myanmar, Vietnam, Nepal, Cambodia) with diaspora in Singapore can also be directly commissioned [(World Bank, n.d.-e)](#ref39). This can be done through mandating the reporting of quote prices directly to the Monetary Authority of Singapore.
## 7.2 Future Works

Due to page constraints, more intensive correlational studies and modelling were not conducted on the individual corridors beyond a panel ([Figure 7](#Figure-7)) that can only be deemed descriptive and not predictive, especially for RQ2. This refers to modelling on series that reflect the destination country's own economic trends, like tariffs or sanctions, interest rates, policy changes, and economic crises like the Sri Lankan 2022 sovereign debt default.

More factors can always be used in RQ3-style analysis, the only constraints are the page limit and data availability.

More corridors from Singapore to other countries with significant diaspora can be analysed, such as Vietnam, Myanmar, Nepal and Cambodia. The data used from the World Bank's databank only has nine corridors within this study's scope, and similar methodology can be used by NGOs or other world organisations to build up a more comprehensive databank. Additionally, if MOM publishes work permit holder statistics by nationality, it will be easier to identify which corridors to focus on.

Other countries with large migrant populations that send substantial remittances (like Qatar, the UAE and Saudi Arabia) can utilise this structure to conduct analysis on remittance outflows from their country specifically. As the B portions of the research questions will be the same, they only need to develop the region- or corridor-specific A structure, and they could conduct a deeper analysis if they were not restricted by length.

Machine-learning models can also be developed (instead of the classical time-series ones used here) for better forecastability.
