# Appendix B — Model tournament framework

This is the full code powering all six model tournaments (both scopes × three constructions) and their chart pairs, imported in the main report at the top of the modelling subsection of 4.3.1 b2. 

```python
# Model tournament: 7 candidates vs a naive baseline - fit, residual validity,
# time-ordered holdout (Ch 8 step 4). Fitted on the RAW quarterly series.
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.stats.diagnostic import acorr_ljungbox

if "APPX_BY" not in globals():  # evidence-trail buffer, created by whichever script runs first
    APPX_SECS, APPX_BY, APPX_CURRENT = [], {}, [None]

    def appx(*args, sep=" "):
        k = APPX_CURRENT[0] or "(preamble)"
        if k not in APPX_BY:
            APPX_BY[k] = []
            APPX_SECS.append(k)
        APPX_BY[k].append(sep.join(str(a) for a in args))

    def appx_mark(t):
        APPX_CURRENT[0] = t
        if t not in APPX_BY:
            APPX_BY[t] = []
            APPX_SECS.append(t)


def trail_section(label):
    if label == "worldwide FX margin":
        return "4.3.2.1 2b"
    if label in ("worldwide fee", "Singapore fee", "Singapore FX margin"):
        return "4.3.2.2 2a"
    return "4.3.1.2 1a" if label.startswith(("SG ", "Singapore")) else "4.3.1.1 1b"


TOURN_TABLES = {}  # evaluation tables, shown in the body where cited


def model_tournament(series, breakpoint, label, holdout=8, show_table=False):
    appx_mark(trail_section(label))
    observed = series.dropna()
    observed.index = np.round(observed.index.values, 2)
    grid = np.round(np.arange(observed.index.min(), observed.index.max() + 1e-9, 0.25), 2)
    filled = observed.reindex(grid)
    gaps = int(filled.isna().sum())
    # Isolated gaps (semi-annual 2011-12 era) imputed as the mean of adjacent quarters;
    # the assert guards the single-gap assumption.
    filled = filled.fillna((filled.shift(1) + filled.shift(-1)) / 2)
    assert not filled.isna().any(), "non-isolated gap - neighbour average invalid"
    t, y = filled.index.values, filled.values
    n = len(y)
    n_train = n - holdout
    y_train, y_test = y[:n_train], y[n_train:]

    def design_linear(times):
        return (times - times[0])[:, None]

    def design_quadratic(times):
        time_centred = times - times[0]
        return np.column_stack([time_centred, time_centred**2])

    def design_hinge(times):
        return np.column_stack([times - times[0], np.maximum(0.0, times - breakpoint)])

    def residual_diagnostics(residuals):
        ac1 = np.corrcoef(residuals[:-1], residuals[1:])[0, 1]
        ljung_box_p = acorr_ljungbox(residuals, lags=[8], return_df=True)["lb_pvalue"].iloc[0]
        return ac1, ljung_box_p

    # The naive baseline is built first so every model can be tested against it.
    naive_fc = np.repeat(y_train[-1], holdout)
    naive_errors = y_test - naive_fc

    def dm_p(forecast):
        from scipy import stats as scipy_stats

        d = (y_test - np.asarray(forecast)) ** 2 - naive_errors**2
        n_diff = len(d)
        variance = ((d - d.mean()) ** 2).sum() / n_diff**2
        if not np.isfinite(variance) or variance <= 0:
            return np.nan
        dm_stat = d.mean() / np.sqrt(variance) * np.sqrt((n_diff - 1) / n_diff)
        return 2 * (1 - scipy_stats.t.cdf(abs(dm_stat), df=n_diff - 1))

    rows = []

    def add_model(name, fit, forecast, skip_first=False):
        residuals = (y - fit)[1:] if skip_first else y - fit
        ac1, ljung_box_p = residual_diagnostics(residuals)
        rows.append(
            (
                name,
                np.sqrt(np.mean(residuals**2)),
                ac1,
                ljung_box_p,
                np.sqrt(np.mean((y_test - forecast) ** 2)),
                dm_p(forecast),
            )
        )

    # three trend models: same OLS recipe, different design matrix
    for name, design_fn in [
        ("linear trend", design_linear),
        ("quadratic trend", design_quadratic),
        ("hinge (1 break)", design_hinge),
    ]:
        add_model(
            name,
            LinearRegression().fit(design_fn(t), y).predict(design_fn(t)),
            LinearRegression().fit(design_fn(t)[:n_train], y_train).predict(design_fn(t)[n_train:]),
        )
    # Both Holt variants: the pair isolates how much the damping assumption contributes.
    for holt_name, damped in [("Holt linear trend", False), ("Holt damped trend", True)]:
        holt_model = ExponentialSmoothing(y, trend="add", damped_trend=damped).fit()
        add_model(
            holt_name,
            holt_model.fittedvalues,
            ExponentialSmoothing(y_train, trend="add", damped_trend=damped).fit().forecast(holdout),
        )
    # ARIMA: AIC picks the memory structure (p, q) on the training window
    order = min(((ARIMA(y_train, order=(p, 1, q)).fit().aic, (p, 1, q)) for p in range(3) for q in range(3)))[
        1
    ]
    arima_name = f"ARIMA{order}"
    add_model(
        arima_name,
        ARIMA(y, order=order).fit().fittedvalues,
        ARIMA(y_train, order=order).fit().forecast(holdout),
        skip_first=True,
    )
    # the hinge trend again, now with AR(1) errors handling the streaks
    add_model(
        "hinge + AR(1) errors",
        SARIMAX(y, exog=design_hinge(t), order=(1, 0, 0), trend="c").fit(disp=False).fittedvalues,
        SARIMAX(y_train, exog=design_hinge(t)[:n_train], order=(1, 0, 0), trend="c")
        .fit(disp=False)
        .forecast(holdout, exog=design_hinge(t)[n_train:]),
        skip_first=True,
    )
    # the baseline every model must beat: next quarter = this quarter
    rows.append(("naive (last value)", np.nan, np.nan, np.nan, np.sqrt(np.mean(naive_errors**2)), np.nan))
    TOURN_TABLES[label] = (
        pd.DataFrame(
            rows,
            columns=["model", "in-sample RMSE", "resid ac1", "Ljung-Box p", "holdout RMSE", "DM p vs naive"],
        )
        .set_index("model")
        .round(3)
    )
    if show_table is not False:  # True shows the table here; a list collects it for side_by_side
        styled_table = (
            TOURN_TABLES[label]
            .astype(object)
            .where(TOURN_TABLES[label].notna(), "-")
            .style.set_caption(label)
        )
        show_table.append(styled_table) if isinstance(show_table, list) else display(styled_table)
    appx(
        f"=== {label}: {n} quarters ({gaps} interpolated) | breakpoint "
        f"{qlabel(breakpoint)} | holdout = last {holdout} quarters ==="
    )
    if not show_table:
        # table rows shown in the body are not duplicated into the appendix
        appx(
            f"{'model':<22}{'in-sample RMSE':>15}{'resid ac1':>11}"
            f"{'Ljung-Box p':>13}{'holdout RMSE':>14}{'DM p vs naive':>15}"
        )
        for row in rows:
            pad = lambda value, width: (f"{value:>{width}.3f}" if value == value else f"{'-':>{width}}")
            appx(
                f"{row[0]:<22}{pad(row[1],15)}{pad(row[2],11)}{pad(row[3],13)}{pad(row[4],14)}{pad(row[5],15)}"
            )

    # Nested-model ANOVA (extra-sum-of-squares F). Dependent errors and searched knots
    # both inflate F, so p-values are optimistic.
    def trend_sse(design_fn, knots=None):
        design = design_fn(t) if knots is None else hinge_design(t, knots)[:, 1:]
        trend_model = LinearRegression().fit(design, y)
        return float(((y - trend_model.predict(design)) ** 2).sum())

    two_knots = search_hinge_knots(t, y, 2)
    sse = {
        "linear": (trend_sse(design_linear), 2),
        "quadratic": (trend_sse(design_quadratic), 3),
        "hinge1": (trend_sse(design_hinge), 4),
        "hinge2": (trend_sse(None, two_knots), 6),
    }
    anova_rows = []
    for a, b, comparison_label in [
        ("linear", "quadratic", "quadratic vs linear"),
        ("linear", "hinge1", "hinge (1 break) vs linear"),
        ("hinge1", "hinge2", "hinge (2 breaks) vs hinge (1)"),
    ]:
        (sse_simple, params_simple), (sse_complex, params_complex) = sse[a], sse[b]
        f_stat = ((sse_simple - sse_complex) / (params_complex - params_simple)) / (
            sse_complex / (n - params_complex)
        )
        p = 1 - stats.f.cdf(f_stat, params_complex - params_simple, n - params_complex)
        anova_rows.append(
            (
                comparison_label,
                sse_simple,
                sse_complex,
                f_stat,
                p,
                "complexity justified" if p < 0.05 else "NOT justified - keep the simpler model",
            )
        )
    appx(
        "\nANOVA - does the added complexity earn its parameters? "
        "(extra-sum-of-squares F-tests, nested trend models)"
    )
    appx(f"{'comparison':<34}{'SSE simple':>11}{'SSE complex':>12}" f"{'F':>9}{'p':>9}   verdict")
    for comparison, sse_simple, sse_complex, f_stat, p, verdict in anova_rows:
        appx(f"{comparison:<34}{sse_simple:>11.3f}{sse_complex:>12.3f}{f_stat:>9.2f}{p:>9.4f}   {verdict}")
    appx(
        f"(hinge-2 knots searched at {qlabel(two_knots[0])} and {qlabel(two_knots[1])}; "
        "F p-values optimistic - see cell comment)"
    )
    # Robustness: refit the decisive models on the fully-quarterly era (2013+), no
    # imputation.
    if gaps or t[0] < 2013.0:
        # 2013+ era contains no interpolated quarters, so this refit is imputation-free
        # whether the input series was pre-interpolated or not.
        mask = t >= 2013.0
        time_2013, y_2013 = t[mask], y[mask]
        n_train_2013 = len(y_2013) - holdout
        design_2013 = np.column_stack([time_2013 - time_2013[0], np.maximum(0.0, time_2013 - breakpoint)])
        forecast_2013 = (
            SARIMAX(y_2013[:n_train_2013], exog=design_2013[:n_train_2013], order=(1, 0, 0), trend="c")
            .fit(disp=False)
            .forecast(holdout, exog=design_2013[n_train_2013:])
        )
        rmse_hinge = np.sqrt(np.mean((y_2013[n_train_2013:] - forecast_2013) ** 2))
        rmse_naive = np.sqrt(np.mean((y_2013[n_train_2013:] - y_2013[n_train_2013 - 1]) ** 2))
        appx(
            f"no-imputation refit (2013+ era, {len(y_2013)} quarters): hinge+AR(1) holdout RMSE {rmse_hinge:.3f} vs naive {rmse_naive:.3f}"
        )
    naive_rmse = rows[-1][4]
    beating = [row for row in rows[:-1] if row[4] < naive_rmse]
    verdict = (
        f"{min(beating, key=lambda row: row[4])[0]} beats the naive baseline"
        if beating
        else "no model beats the naive baseline"
    )
    appx(f"verdict: {verdict}")
```

**Claude:**

These fall into three groups: benchmark models, trend models, and diagnostic tests. Here's each one and how they relate.

#### Benchmark models

**Naive.** Tomorrow's forecast equals today's value: ŷ(t+h) = y(t). No parameters, nothing estimated. It exists to be beaten — if a fancier model can't out-forecast "assume nothing changes," the fancier model isn't earning its complexity. For seasonal data the seasonal naive (same value as one season ago) is the usual benchmark instead.

**Linear trend.** Regress y on time: y(t) = a + b·t + error. Assumes growth is a fixed amount per period, forever. Simple and interpretable, but rigid — it can't bend if the series accelerates, slows, or changes direction.

**Quadratic trend.** Adds a t² term: y(t) = a + b·t + c·t² + error. Lets the trend curve. The danger is extrapolation: a parabola fitted to the past keeps curving the same way, so forecasts can shoot up or crash implausibly a few periods out. Usually fine for short horizons, risky for long ones.

#### Hinge (piecewise linear) models

**Hinge.** A linear trend with a kink at a chosen breakpoint k: y(t) = a + b·t + c·max(0, t − k) + error. Before k the slope is b; after k it's b + c. This is the natural choice when something structural changed at a known date (a policy change, a reopening, a pricing shift) and you want the post-break slope to drive the forecast rather than the whole history. You pick k by prior knowledge or by searching for the breakpoint that minimises residual error.

**Hinge with AR(1) errors.** Same hinge mean, but the error term follows e(t) = φ·e(t−1) + white noise. Trend regressions on time series almost always leave autocorrelated residuals — this period's deviation from trend predicts next period's. Ignoring that gives you misleadingly tight standard errors and forecasts that don't use the information in the latest residual. Modelling the AR(1) structure fixes both: coefficient tests become honest, and short-term forecasts get a nudge in the direction of the most recent surprise. If the Ljung-Box test (below) rejects on the plain hinge, this is the first repair to try.

#### Exponential smoothing (Holt)

**Holt linear trend.** Two smoothed components — level and slope — updated each period with smoothing weights α and β. Unlike the regression trends above, the slope is *local*: it adapts to recent data rather than being fixed over the whole sample. Forecasts extend the current slope in a straight line. Good when the trend direction is stable but its magnitude drifts.

**Holt damped trend.** Adds a damping parameter φ (0 < φ < 1) so the slope shrinks geometrically with each step ahead: the forecast flattens toward a plateau instead of growing forever. Empirically this is one of the most reliable general-purpose methods because real series rarely sustain a constant trend indefinitely. It's often the strongest single competitor in forecasting comparisons.

#### ARIMA

ARIMA(p, d, q) models the series as a function of its own past values (AR, order p) and past forecast errors (MA, order q), after differencing d times to remove non-stationarity. It's the most flexible of the set — it can mimic a random walk, a trend, mean reversion, or cycles depending on the orders chosen. Orders are usually picked by an automated search over AIC/BIC (auto.arima or equivalent). Its weakness is interpretability and a tendency to produce flat or mean-reverting long-horizon forecasts after differencing. Seasonal data needs the SARIMA extension.

#### Diagnostic tests

**Ljung-Box test.** Tests whether a model's residuals are white noise. Null: no autocorrelation up to lag h. If you reject (small p-value), the model has left predictable structure in the residuals — it's under-specified, and there's forecastable information being thrown away. Apply it to every fitted model; a model that fails is a candidate for AR errors, higher ARIMA orders, or seasonal terms. Note it's a check on *adequacy*, not accuracy: a model can pass Ljung-Box and still forecast poorly.

**Diebold-Mariano test.** Compares the out-of-sample forecast accuracy of two models directly. You compute the loss (squared or absolute error) from each model's forecasts over a holdout period, take the difference series, and test whether its mean is zero. Null: equal accuracy. Rejecting tells you one model is statistically better, not just better by luck of the sample. The test accounts for autocorrelation in the loss differences, which matters for multi-step forecasts. Use the Harvey-Leybourne-Newbold small-sample correction if your holdout is short. This is how you decide whether, say, Holt damped beating ARIMA by 3% RMSE is a real finding or noise.

#### How these fit together in a study

The usual workflow is: fit everything on a training window, check each with Ljung-Box, generate rolling or fixed-origin forecasts on a holdout, compare accuracy metrics (RMSE, MAE, MAPE), then use Diebold-Mariano on the leading pairs to establish which differences are significant. Naive anchors the bottom; the hinge models earn their place only if a structural break is plausible in your data.

**Claude:**

A modelling workflow is a loop, not a line — you'll cycle back through earlier stages as later ones reveal problems. But here's the canonical sequence and what each stage is for.

#### 1. Define the question

Before touching data, pin down three things: what you're trying to predict or explain (the target), what decision the answer feeds, and how you'll know the model is good enough. "Good enough" means a concrete metric and a threshold — forecast error under X, classification accuracy above Y, a coefficient estimated precisely enough to be useful. Skipping this is the most common cause of models that are technically fine and practically useless.

#### 2. Get and understand the data

Collect the data, then spend real time on exploratory analysis: distributions, missing values, outliers, time coverage, how variables relate to each other and to the target. Plot everything. This stage tells you which models are even plausible — a target that's skewed, bounded, or seasonal rules out some options and suggests others. You're also building the intuition you'll need later to judge whether model outputs make sense.

#### 3. Clean and prepare

Handle missing data (drop, impute, or model it), fix inconsistencies, transform variables where needed (logs for skewed data, differencing for non-stationary series, encoding for categories). Engineer features if the model will benefit from derived variables. Every choice here should be recorded — these decisions affect results as much as the model choice does, and you'll need to reproduce them exactly on new data.

#### 4. Split the data

Separate the data you'll fit on from the data you'll evaluate on *before* fitting anything. For cross-sectional data that's a random train/test split, often with cross-validation inside the training portion. For time series it must be chronological — train on the past, test on the future — because random splits leak future information backward and inflate performance. Set the test set aside and don't look at it until the end.

#### 5. Establish a baseline

Fit the simplest defensible model first: the mean, the last value, a plain linear regression. This is your floor. Every subsequent model must beat it meaningfully, or the added complexity isn't earning its keep. Baselines also catch pipeline bugs early — if a sophisticated model can't beat "predict the average," something is wrong with the setup, not the model.

#### 6. Fit candidate models

Try a range of model families appropriate to the problem, from simple to complex. Fit each on the training data only. Tune hyperparameters using cross-validation or a validation subset — never the test set. Keep the set of candidates manageable; a handful of well-chosen models beats a scattershot of twenty.

#### 7. Diagnose

For each fitted model, check that its assumptions hold and that it has extracted the available signal. Residual plots, autocorrelation tests, checks for heteroskedasticity, influence of outliers, stability across subsets. A model that fails diagnostics is misspecified, and its performance numbers can't be trusted even if they look good. Fix and refit before moving on.

#### 8. Evaluate and compare

Now use the held-out test data. Score every candidate on the metric you fixed in step 1. Compare them to the baseline and to each other, and where the differences are small, test whether they're statistically distinguishable rather than noise. Look beyond the headline number: does the model fail in particular regions, periods, or subgroups? Is error concentrated where it matters most?

#### 9. Select and interpret

Choose the model — not necessarily the best-scoring one. Weigh accuracy against interpretability, stability, data requirements, and how it will be maintained. Then interpret it: what are the drivers, what's the uncertainty, where should it not be trusted? Communicate limitations as clearly as results.

#### 10. Deploy, monitor, revisit

If the model is used operationally, put it into production with the same preprocessing applied identically, then monitor its performance over time. Data drifts, relationships change, and models decay. Set a trigger for when to retrain or revisit from step 1.

#### The principles underneath

Three rules hold the whole thing together. Keep evaluation data untouched until the end, or your performance estimates are fiction. Prefer the simplest model that does the job, because complexity costs interpretability and robustness. And document every decision, because a model you can't reproduce is a model you can't defend.
